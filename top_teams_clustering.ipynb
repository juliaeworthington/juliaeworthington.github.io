```python
# importing packages 

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.preprocessing import StandardScaler
from scipy.cluster.hierarchy import dendrogram, linkage
```

```python
# importing data

df = pd.read_csv(f"data/top_teams.csv")
X1 = df['Maps']
X2 = df['Rating3.0']
plt.scatter(X1, X2)
plt.title('Original Data')
plt.xlabel('Number of Maps against Top 20 Teams')
plt.ylabel('Rating3.0')
plt.savefig('maps_v_rating_scatter.png')
plt.show()
```

```python
# standardizing data

data = df.drop(columns=['Team', 'K-D Diff', 'K/D'])
X = StandardScaler().fit_transform(data)
plt.scatter(X[:,0], X[:,1])
plt.title('Standardized Data')
plt.xlabel('Number of Maps against Top 20 Teams (standardized)')
plt.ylabel('Rating3.0 (standardized)')
plt.savefig('maps_v_rating_scatter_standard.png')
plt.show()
```

```python
silhouette_scores = [] 
for i in range(2,15):
    km = KMeans(n_clusters = i)
    score = silhouette_score(X, km.fit_predict(X))
    silhouette_scores.append(score)
x = range(2,15)
y = silhouette_scores

plt.plot(x, y, marker = 'o', markerfacecolor = 'r')
plt.xlabel('Number of Clusters')
plt.ylabel('Silhouette Score')
plt.title('Silhouette Plot')
plt.savefig('maps_v_rating_silhouette.png')
plt.show()
```

```python
km = KMeans(n_clusters = 6)
km.fit(X)
plot_data = np.array([km.labels_, X1, X2]).T
labels = km.labels_
for i, colr in zip(np.unique(labels), ["#2C5F8A", "#B03A2E", "#3E7C5A", "#BFAC4C", "#111111", "#9F2B68"]):
    lab = labels == i
    plt.scatter(plot_data[lab, 1], plot_data[lab, 2], alpha=0.85, color=colr)
plt.title('Visualizing the Clusters in 2D k=6')
plt.xlabel('Number of Maps against Top 20 Teams')
plt.ylabel('Rating3.0')
plt.savefig('teams_6_clusters.png')
plt.show()
```
